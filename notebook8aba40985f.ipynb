## 🌐 Cell 2 — Install Brave Browser
> Installs via official `.deb` — no `snap` required.

```python
import subprocess
subprocess.run('curl -fsS https://dl.brave.com/install.sh | sh', shell=True)

print('✅ Brave installed!')
```

```python
import subprocess
import shutil

def run_command(command, description):
    print(f"🔄 Installing {description}...")
    try:
        subprocess.run(command, shell=True, check=True)
        print(f"✅ {description} installed successfully!\n")
    except subprocess.CalledProcessError:
        print(f"❌ Failed to install {description}.\n")

def main():
    # 1. Install Brave Browser (Your original code)
    run_command('curl -fsS https://brave.com | sh', 'Brave Browser')

    # 2. Prepare system for 32-bit architecture (Crucial for Steam & Wine)
    print("🔧 Configuring 32-bit architecture support...")
    subprocess.run('sudo dpkg --add-architecture i386', shell=True)
    run_command('sudo apt-get update', 'package lists update')

    # 3. Install Steam & Wine natively
    run_command('sudo apt-get install -y wine steam', 'Wine and Steam')

    # 4. Install Flatpak & Flathub (Required to install Sober)
    if not shutil.which('flatpak'):
        run_command('sudo apt-get install -y flatpak', 'Flatpak core')
    
    # Add Flathub repository if it doesn't exist
    subprocess.run('flatpak remote-add --if-not-exists flathub https://flathub.org', shell=True)

    # 5. Install Sober (Roblox wrapper) via Flathub
    # Note: The precise Flathub ID for Sober is org.vinegarhq.Sober
    run_command('flatpak install -y flathub org.vinegarhq.Sober', 'Sober (Roblox)')

    print("🎉 Setup complete! Brave, Steam, Wine, and Sober are installed.")

if __name__ == '__main__':
    main()
```

## 🖥️ Cell 3 — Start Virtual Display + XFCE4

```python
import subprocess, os, time

os.environ['DISPLAY'] = ':99'

subprocess.Popen('Xvfb :99 -screen 0 1280x800x24 -ac', shell=True,
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(3)
print('✅ Xvfb running!')

subprocess.Popen('DISPLAY=:99 startxfce4', shell=True,
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(6)
print('✅ XFCE4 running!')
```

## 📡 Cell 4 — Start VNC Server

```python
import subprocess, time

VNC_PASSWORD = '12345678'  # Change if you want

subprocess.run(f'x11vnc -storepasswd {VNC_PASSWORD} /root/.vncpasswd', shell=True,
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

subprocess.Popen(
    'x11vnc -display :99 -rfbauth /root/.vncpasswd -rfbport 5900 -forever -loop -shared -noxdamage',
    shell=True, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(2)
print(f'✅ VNC server running (password: {VNC_PASSWORD})')
```

## 🖱️ Cell 5 — Start noVNC

```python
import subprocess, time
subprocess.Popen('websockify --web=/usr/share/novnc/ 6080 localhost:5900', shell=True,
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
time.sleep(2)
print('✅ noVNC running on port 6080!')
```

## 🔗 Cell 6 — Create Public Tunnel via localhost.run
> ✅ Free | ✅ No account | ✅ No credit card | Uses SSH only

```python
import subprocess, time

subprocess.Popen(
    'ssh -o StrictHostKeyChecking=no -o ServerAliveInterval=60 '
    '-R 80:localhost:6080 nokey@localhost.run '
    '> /kaggle/working/tunnel.log 2>&1',
    shell=True
)

print('⏳ Starting tunnel...')
time.sleep(15)

with open('/kaggle/working/tunnel.log', 'r') as f:
    content = f.read()

found = False
for line in content.split('\n'):
    if 'localhost.run' in line and 'https://' in line:
        for word in line.split():
            if (word.startswith('https://') and 'localhost.run' in word
                    and not any(x in word for x in ['admin.', '/docs'])):
                print('=' * 55)
                print('🎉 Your remote desktop is ready!')
                print(f'🌐 Open: {word.strip()}/vnc.html')
                print(f'🔒 Password: {VNC_PASSWORD}')
                print('=' * 55)
                found = True
                break

if not found:
    print('⚠️ URL not ready — run Cell 7 to check the log')
```
